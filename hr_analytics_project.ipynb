```python
import pandas as pd

df = pd.read_csv("HR_Analytics.csv")
df.head()
```

Data Cleaning

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
df.dropna(inplace=True)
```



*   Experience Group



```python
print(df.columns)
```

```python
df['experience_group'] = pd.cut(
    df['yearsatcompany'],
    bins=[0,2,5,10,40],
    labels=['0-2 Years','3-5 Years','6-10 Years','10+ Years']
)
```



*   Salary Band



```python
df['salary_band'] = pd.qcut(
    df['monthlyincome'],
    q=4,
    labels=['Low','Medium','High','Very High']
)
```



*   Attrition Flag



```python
df['attrition_flag'] = df['attrition'].map({'Yes':1,'No':0})
```



*   sanity check



```python
print(df[['yearsatcompany','experience_group']].head())
print(df[['monthlyincome','salary_band']].head())
```

```python
print(df.groupby('department')['attrition_flag'].mean())
print(df.groupby('salary_band')['attrition_flag'].mean())
print(df.groupby('experience_group')['attrition_flag'].mean())
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
df = df.rename(columns={
    'employeenumber': 'employee_id',
    'monthlyincome': 'monthly_income',
    'yearsatcompany': 'years_at_company'
})
```

```python
df['experience_group'] = pd.cut(
    df['years_at_company'],
    bins=[0,2,5,10,40],
    labels=['0-2 Years','3-5 Years','6-10 Years','10+ Years']
)
```

```python
df['salary_band'] = pd.qcut(
    df['monthly_income'],
    q=4,
    labels=['Low','Medium','High','Very High']
)
```

```python
print(df[['employee_id','monthly_income','years_at_company']].head())
```

```python
print(df.groupby('department')['attrition_flag'].mean())
```

```python
df.to_csv("cleaned_hr_data.csv", index=False)
```
